```python
#write a function get _grade(percent)that returns "A"(>=85), "B"(>=70), "C"(>=50), or "D" 
percent = float(input("Enter your percentage: "))
print("Your grade is:", get_grade(percent))
def get_grade(percent):
    if percent >= 85:
        return "A"
    elif percent >= 70:
        return "B"
    elif percent >= 50:
        return "C"
    else:
        return "D"
```

```python
def add_all(*args):
    total = 0
    for num in args:
        total += num
    return total

print(add_all(5, 10))             # Output: 15
print(add_all(5, 10, 20, 30, 50)) # Output: 115
```

```python
def calculate_bill(amount,discount=1,tax=5):
    after_discount =amount- amount * discount/100
    return  after_discount + after_discount * tax / 100

print(calculate_bill(2000))
print(calculate_bill(2000, discount=10))
print(calculate_bill(2000, discount=20, tax=18))
```
